```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification
```

```python
X = pd.read_csv('dataset/UNSW_NB15_training-set.csv', na_values='-')
y = X[['label']]

X.drop(columns=['label', 'id', 'attack_cat'], inplace=True)
```

```python
print(f'X Shape: {X.shape}')
print(f'y Shape: {y.shape}')
```

```python
from sklearn.preprocessing import StandardScaler
X = pd.get_dummies(X, columns=['state', 'service','proto'], drop_first=True)

scaler = StandardScaler()
X = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
```

```python
print(X.shape, y.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from matplotlib import pyplot

model = LogisticRegression()
model.fit(X, y)
importance = model.coef_[0]
for i, v in enumerate(importance):
    print(f'Feature: {X.columns[i]}, Score: {v}')

pyplot.bar([x for x in range(len(importance))], importance)
pyplot.show()
```

```python
feature_names = X.columns
values = model.coef_[0]
abs_values = np.abs(values)

features = pd.DataFrame({
    'feature': feature_names,
    'values': values,
    'abs_values': abs_values
})
pd.set_option('display.max_rows', None)
features = features.sort_values('abs_values', ascending=False)
features
```
